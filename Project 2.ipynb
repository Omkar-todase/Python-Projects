# Project 2
- Task 1: Counting the number of each nucleotides and demonstrate in the bar plot
- Task 2: Counting the number of proteins and demonstrate in the bar plot

```python
from Bio import SeqIO
import Bio.Seq
import matplotlib.pyplot as plt
import seaborn as sns

for record in Bio.SeqIO.parse('Eubacterium ventriosum.fasta', 'fasta'):
    Eubacterium_DNA = record.seq

# Task 1: Counting the number of each nuelcotides and plot a bar graph 

count_nucleotides = {
    'A': Eubacterium_DNA.count('A'),
    'T': Eubacterium_DNA.count('T'),
    'C': Eubacterium_DNA.count('C'),
    'G': Eubacterium_DNA.count('G')
}
print(count_nucleotides)
print(' ')


plt.figure(figsize= (10, 5))
plt.bar(count_nucleotides.keys(), count_nucleotides.values(), width = 0.4, color = ['blue', 'red', 'green', 'purple'])
plt.title('Nucleotides Frequency', size = 10, weight = 'bold')
plt.xlabel('Nucleotides', size = 10, weight = 'bold')
plt.ylabel('Frequency', size = 10, weight = 'bold')
plt.show()
```

```python
# Task 2: Counting the Number of proteins

Eubacterium_mRNA = Eubacterium_DNA.transcribe()
Eubacterium_AA = Eubacterium_mRNA.translate()

from collections import Counter
Proteins = Counter(Eubacterium_AA)
print(f'Number of proteins :{Proteins}')
print('    ')

plt.figure(figsize= (10, 5))
plt.bar(Proteins.keys(), Proteins.values(), width = 0.4, color = ['blue', 'red', 'green', 'purple'])
plt.title('Protein Sequence Frequency', size = 10, weight = 'bold')
plt.xlabel('Number of Proteins', size = 10, weight = 'bold')
plt.ylabel('Frequency', size = 10, weight = 'bold')
plt.show()
```
