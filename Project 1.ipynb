# Project 1
- Task 1: Retriving the orchid fasta file from Github.
- Task 2: Open that fasta file in jupyter notebook
- Task 3: Find out total number of genes and its highest GC content gene and Lowest GC conent gene along with its ID
- Task 4: Find out its highest Molecular Weight gene and Lowest Molecular Weight gene along with its ID

```python
# Task 1: Retriving the orchid fasta file from Github
import urllib.request

url = "https://raw.githubusercontent.com/biopython/biopython/master/Doc/examples/ls_orchid.fasta"

urllib.request.urlretrieve(url, "ls_orchid.fasta")
```

```python
# Task 2: Open the fasta file in jupyter notebook
from Bio import SeqIO
import Bio.Seq
from Bio.SeqUtils import gc_fraction

total_genes = 0
Highest_GC_content = 0
Highest_gene_ID = None

Lowest_GC_content = float('inf')   
Lowest_GC_gene_ID = None

#  Task 3:total number of genes and Highest_GC_Content  
for record in Bio.SeqIO.parse('ls_orchid.fasta', 'fasta'):
    total_genes += 1
    gene_length = len(record.seq)

    gc_content = round(gc_fraction(record.seq), 2)

    if gc_content > Highest_GC_content:
        Highest_GC_content = gc_content
        Highest_gene_ID = record.id
print(f"Total number of genes: {total_genes}")
print(f' The gene ID : {Highest_gene_ID}: Highest GC content {Highest_GC_content} %')
print('')

# Task: 3 Lowest_GC_Content
for record in Bio.SeqIO.parse('ls_orchid.fasta', 'fasta'):
    gc_content = round(gc_fraction(record.seq), 2)

    if gc_content < Lowest_GC_content:
        Lowest_GC_content = gc_content
        Lowest_gene_ID = record.id
print(f' The gene ID : {Lowest_gene_ID}: Lowest GC content {Lowest_GC_content} %')
```

```python
from Bio.SeqUtils import molecular_weight

Highest_Mol_weight = 0
Highest_gene_ID = None

Lowest_Mol_weight = float('inf')   
Lowest_gene_ID = None

# Task :4 Highest Molecular weight
for record in Bio.SeqIO.parse('ls_orchid.fasta', 'fasta'):
    gene_length = record.seq
    clean_seq = gene_length.replace('N', '')

    mol_weight = round(molecular_weight(clean_seq, 'DNA'), 2)
    
    if mol_weight > Highest_Mol_weight:
        Highest_Mol_weight = mol_weight
        Highest_gene_ID = record.id
print(f' The gene ID : {Highest_gene_ID}: Highest Mol Weight {Highest_Mol_weight} %')
print('')

# Task :4 Lowest Molecular Weight
for record in Bio.SeqIO.parse('ls_orchid.fasta', 'fasta'):
    gene_length = record.seq
    clean_seq = gene_length.replace('N', '')

    if mol_weight < Lowest_Mol_weight:
        Lowest_Mol_weight = mol_weight
        Lowest_gene_ID = record.id 
print(f' The gene ID : {Lowest_gene_ID}: Highest Mol Weight {Lowest_Mol_weight} %')
```
